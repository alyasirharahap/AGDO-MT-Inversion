```python
import numpy as np

from yasir_agdo_mt.core.forward import (
    mt1d,
    apparent_resistivity,
    phase,
)

from yasir_agdo_mt.core.objective import rmse_mt
```

```python
def misfit(rho_obs, ph_obs, rho_cal, ph_cal):

    m = np.sqrt(np.mean(
        (np.log10(rho_obs/rho_cal))**2 +
        (np.deg2rad(ph_obs-ph_cal))**2
    ))

    return m
```

```python
freq = np.logspace(-3, 4, 56)

rho = np.array([100, 10, 500])

thick = np.array([50, 100])
```

```python
Z = mt1d(
    rho,
    thick,
    freq
)

rho_obs = apparent_resistivity(
    Z,
    freq
)

phase_obs = phase(
    Z
)
```

```python
rho_cal = rho_obs.copy()
phase_cal = phase_obs.copy()

rho_cal *= 1.02
phase_cal += 0.3
```

```python
rmse_notebook = misfit(
    rho_obs,
    phase_obs,
    rho_cal,
    phase_cal,
)

rmse_library = rmse_mt(
    rho_obs,
    phase_obs,
    rho_cal,
    phase_cal,
)
```

```python
print("=" * 50)
print("OBJECTIVE FUNCTION VALIDATION")
print("=" * 50)

print(f"Notebook RMSE : {rmse_notebook:.15f}")
print(f"Library RMSE  : {rmse_library:.15f}")

print()

print(f"Match         : {np.isclose(rmse_notebook, rmse_library, atol=1e-12)}")
print(f"Difference    : {abs(rmse_notebook-rmse_library):.15e}")
```
