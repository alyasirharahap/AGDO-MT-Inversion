```python
import numpy as np

from yasir_agdo_mt.core.forward import (
    mt1d,
    apparent_resistivity,
    phase,
)

from yasir_agdo_mt.core.objective import (
    rmse_mt,
    evaluate_mt_model,
)
```

```python
def misfit(rho_obs, ph_obs, rho_cal, ph_cal):

    m = np.sqrt(np.mean(
        (np.log10(rho_obs/rho_cal))**2 +
        (np.deg2rad(ph_obs-ph_cal))**2
    ))

    return m
```

```python
freq = np.logspace(-3, 4, 56)

rho = np.array([100, 10, 500])

thick = np.array([50, 100])

model = np.concatenate((rho, thick))
```

```python
Z = mt1d(
    rho,
    thick,
    freq,
)

rho_obs = apparent_resistivity(
    Z,
    freq,
)

phase_obs = phase(Z)
```

```python
model_test = np.array([
    102,
    11,
    490,
    55,
    95
], dtype=float)
```

```python
rho_test = model_test[:3]

thick_test = model_test[3:]

Z = mt1d(
    rho_test,
    thick_test,
    freq,
)

rho_cal = apparent_resistivity(
    Z,
    freq,
)

phase_cal = phase(Z)

fitness_old = misfit(
    rho_obs,
    phase_obs,
    rho_cal,
    phase_cal,
)

fitness_new = evaluate_mt_model(
    model_test,
    freq,
    rho_obs,
    phase_obs,
)
```

```python
print("=" * 50)
print("MODEL EVALUATION VALIDATION")
print("=" * 50)

print(f"Notebook : {fitness_old:.15f}")
print(f"Library  : {fitness_new:.15f}")

print()

print("Match :", np.isclose(
    fitness_old,
    fitness_new,
    atol=1e-12,
))

print("Difference :", abs(
    fitness_old -
    fitness_new
))
```
